```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from IPython.display import display

# Enter your uploaded CSV file path here
FILE_PATH = "/content/Housing.csv"

# Load dataset
df = pd.read_csv(FILE_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("First five records:")
display(df.head())

print("\nLast five records:")
display(df.tail())

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

### Dataset Description

The Housing Prices dataset is obtained from Kaggle. It contains information about residential properties, including their area, number of bedrooms, bathrooms, stories, parking facilities and prices.

For this activity, the `area` column is selected as the independent variable (input), and the `price` column is selected as the dependent variable (target).

The objective is to develop a Simple Linear Regression model to predict house prices based on their area.

**Dataset Name:** Housing Prices Dataset

**Source:** [Kaggle – Housing Prices Dataset](https://www.kaggle.com/datasets/yasserh/housing-prices-dataset)

**File Type:** CSV

```python
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)

print("\nPossible input feature: area")
print("Target variable: price")
```

```python
print("Numerical summary:")
display(df.describe())

print("\nCategorical summary:")
display(df.describe(include="object"))
```

```python
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage.round(2)
})

display(missing_table)

print("Total missing values:", df.isnull().sum().sum())
```

```python
# Create a copy of the original dataset
cleaned_df = df.copy()

# Clean categorical text
for column in cleaned_df.select_dtypes(include="object").columns:
    cleaned_df[column] = (
        cleaned_df[column]
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

# Convert area and price into numeric values
cleaned_df["area"] = pd.to_numeric(
    cleaned_df["area"], errors="coerce"
)

cleaned_df["price"] = pd.to_numeric(
    cleaned_df["price"], errors="coerce"
)

# Remove rows where area or price is missing
before_invalid_removal = len(cleaned_df)

cleaned_df = cleaned_df.dropna(
    subset=["area", "price"]
)

print("Rows removed due to missing area or price:",
      before_invalid_removal - len(cleaned_df))

# Remove invalid non-positive area or price values
before_invalid_values = len(cleaned_df)

cleaned_df = cleaned_df[
    (cleaned_df["area"] > 0) &
    (cleaned_df["price"] > 0)
].copy()

print("Rows removed due to invalid area or price:",
      before_invalid_values - len(cleaned_df))

# Fill remaining numerical missing values using median
for column in cleaned_df.select_dtypes(include=np.number).columns:
    cleaned_df[column] = cleaned_df[column].fillna(
        cleaned_df[column].median()
    )

# Fill remaining categorical missing values using mode
for column in cleaned_df.select_dtypes(include="object").columns:
    if cleaned_df[column].isnull().any():
        mode_value = cleaned_df[column].mode()

        if not mode_value.empty:
            cleaned_df[column] = cleaned_df[column].fillna(
                mode_value[0]
            )

# Check and remove duplicate records
duplicate_count = cleaned_df.duplicated().sum()

print("Duplicate records found:", duplicate_count)

cleaned_df = cleaned_df.drop_duplicates()

print("\nDataset after cleaning:", cleaned_df.shape)
print("Remaining missing values:",
      cleaned_df.isnull().sum().sum())

display(cleaned_df.head())
```

```python
# Independent variable (input)
X = cleaned_df[["area"]]

# Dependent variable (target)
y = cleaned_df["price"]

print("Input feature:")
display(X.head())

print("\nTarget variable:")
display(y.head())

print("\nInput shape:", X.shape)
print("Target shape:", y.shape)
```

### Selection of Input and Target Variables

The `area` column is selected as the independent variable (X), as it represents the size of the house. The `price` column is selected as the dependent variable (Y), as it represents the house price.

Simple Linear Regression is used because only one independent variable is considered for predicting the target variable.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=cleaned_df,
    x="area",
    y="price"
)

plt.title("House Area vs House Price")
plt.xlabel("Area (square feet)")
plt.ylabel("House Price")

plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("Training percentage:",
      round(len(X_train) / len(X) * 100, 2), "%")

print("Testing percentage:",
      round(len(X_test) / len(X) * 100, 2), "%")
```

### Train-Test Split

The dataset is divided into two parts using an 80:20 ratio. 80% of the records are used to train the Linear Regression model, while the remaining 20% are used to test its prediction performance. A random state of 42 is used to make the split reproducible.

```python
model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")

print("Coefficient (slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
slope = model.coef_[0]
intercept = model.intercept_

print("Simple Linear Regression Equation:")
print(f"Price = {slope:.2f} × Area + {intercept:.2f}")

print("\nInterpretation:")
print(
    f"For every one-unit increase in area, "
    f"the predicted house price changes by approximately "
    f"{slope:.2f} price units."
)
```

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

display(results.head(10))
```

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION METRICS")

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R-squared (R2):", round(r2, 4))
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred)

# Perfect prediction reference line
minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.title("Actual vs Predicted House Prices")
plt.xlabel("Actual House Prices")
plt.ylabel("Predicted House Prices")

plt.show()
```

```python
# Select a new area within the training range
new_area = float(X_train["area"].median())

# Predict house price
new_prediction = model.predict(
    pd.DataFrame({"area": [new_area]})
)

print("New house area:", new_area, "square feet")

print("Predicted house price:",
      round(new_prediction[0], 2))
```

```python
print("CONCLUSION")

print(
    f"The dataset originally contained {df.shape[0]} rows "
    f"and {df.shape[1]} columns."
)

print(
    f"After cleaning, {cleaned_df.shape[0]} records "
    f"were available for modelling."
)

print(
    f"The model was trained using {len(X_train)} records "
    f"and tested using {len(X_test)} records."
)

print(
    f"The regression coefficient is {slope:.2f} "
    f"and the intercept is {intercept:.2f}."
)

print(f"MAE: {mae:.2f}")
print(f"MSE: {mse:.2f}")
print(f"R2 Score: {r2:.4f}")

print(
    "\nA Simple Linear Regression model was developed "
    "to predict house prices using house area."
)

print(
    "The model's performance is evaluated using "
    "MAE, MSE, RMSE and R2 score."
)

print(
    "The model uses only one feature, so other factors "
    "such as location, bedrooms and facilities are not considered."
)
```
